```python
class GeneradorCifradoBits:
    def __init__(self, semilla_secreta: int):
        """
        Inicializa el estado interno combinando la semilla con constantes matemáticas
        en un rango estricto de 32 bits.
        """
        # Máscara para simular un registro de 32 bits sin signo (0 a 4,294,967,295)
        self.MASK32 = 0xFFFFFFFF
        
        # Inicialización del estado interno (X, Y, Z)
        # Se aplica un XOR inicial con constantes fijas para aumentar la entropía
        self.x = (semilla_secreta ^ 0x61657870) & self.MASK32
        self.y = (semilla_secreta ^ 0x6e643262) & self.MASK32
        self.z = (semilla_secreta ^ 0x3a6b643a) & self.MASK32

    def siguiente(self) -> int:
        """
        Avanza el estado interno mediante operaciones ARX (Add-Rotate-XOR) 
        y devuelve un número aleatorio de bits mezclados (cifrados).
        """
        # 1. Mutación del Estado (Cambio de estado no lineal)
        # Cada paso arrastra y modifica los valores anteriores aplicando desbordamiento estricto
        self.x = ((self.x ^ (self.y << 5)) + self.z) & self.MASK32
        self.y = ((self.y ^ (self.z >> 3)) + self.x) & self.MASK32
        self.z = ((self.z ^ (self.x << 7)) + self.y) & self.MASK32

        # 2. Cifrado / Mezcla de bits final (Bit Scrambling)
        # Oculta el estado real mezclando los tres componentes con XOR y desplazamientos
        salida_cifrada = self.x ^ self.y
        salida_cifrada = (salida_cifrada ^ (self.z >> 16)) & self.MASK32

        return salida_cifrada

    def consultar_estado(self) -> dict:
        """Devuelve una copia del estado interno actual."""
        return {"x": self.x, "y": self.y, "z": self.z}


# --- EJEMPLO DE USO ---
if __name__ == "__main__":
    # Inicializar el generador con una semilla numérica única de tu elección
    semilla = 987654321
    generador = GeneradorCifradoBits(semilla_secreta=semilla)

    print("--- AVANCE DE ESTADOS EN PYTHON ---")
    print(f"Semilla inicial: {semilla}\n")

    # Generar una secuencia de 5 números pseudoaleatorios criptográficos
    for paso in range(1, 6):
        num_aleatorio = generador.siguiente()
        estado_actual = generador.consultar_estado()
        
        print(f"Paso {paso}:")
        print(f"  -> Número generado (bits cifrados): {num_aleatorio}")
        print(f"  -> Estado interno [X]: {estado_actual['x']}, [Y]: {estado_actual['y']}, [Z]: {estado_actual['z']}")
        print("-" * 50)
```
